## Fixed Size Chunking:

```python
text= """This is the simple test """


def fixed_size_chunking(text, chunk_size=45):
    return [text[i:i+chunk_size] for i in range(0, len(text), chunk_size)]

chunks_fixed = fixed_size_chunking(text=text)
print(chunks_fixed[0])
```

## Sentence Splitting:

```python
import nltk
nltk.download("punkt")
from nltk.tokenize import sent_tokenize


def sentence_splitting(text, max_tokens=30):
    sentences = sent_tokenize(text)
    chunks = []
    current_chunk = []
    current_length = 0

    for sentence in sentences:
        if current_length + len(sentence.split()) <= max_tokens:
            current_chunk.append(sentence)
            current_length += len(sentence.split())
        else:
            chunks.append(' '.join(current_chunk))
            current_chunk = [sentence]
            current_length += len(sentence.split())

    if current_chunk:
        chunks.append(' '.join(current_chunk))

    return chunks


chunks_sentence = sentence_splitting(text)
print(chunks_sentence[0])
```

## Recursive chunking:

```python
def recursive_chunking(text, max_tokens=30):
    if len(text.split()) <= max_tokens:
        return[text]
    else:
        midpoint = len(text) // 2
        first_half = text[:midpoint]
        second_half = text[midpoint:]
        return recursive_chunking(first_half, max_tokens=max_tokens) + recursive_chunking(second_half, max_tokens)
    

chunk_recursive = recursive_chunking(text)
print(chunk_recursive)
```
