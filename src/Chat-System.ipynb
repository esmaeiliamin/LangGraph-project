```python
from openai import OpenAI
```

```python
client = OpenAI() #reads the .env file

def get_response(messages):
    completion = client.chat.completions.create(
        messages = messages,
        model = "gpt-4o-mini",
        max_tokens = 150,
        temperature = 0.5
    )
    return completion.choices[0].message.content.strip()
print("Welcom to your prograamming assistant")
print("Type 'exit' to quite. ")

conversation = []
conversation.append(
    {"role": "system",
     "content": """Respond code only if possible.
     keepe explanations to a minimum
     (2 paragraphs max). Dont use markdown"""}
)
while True:
    prompt = input("You: ")
    if prompt.lower() == 'exit':
        print("Gooodbye")
        break

    #Add prompt to the chat history
    conversation.append({"role": "user", "content": prompt})
    #Process prompt 
    response = get_response(conversation)
    print("AI:")
    print(response)
    print()
    #Add response to chat hisory 
    conversation.append({"role": "assistant",
                         "content": response})
```
