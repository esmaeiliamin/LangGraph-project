# **NER Application:**

#### NER is the process of extracting certain named entities line name 
#### of places, persons or organization from the given text.

```python
import ollama
```

```python
def ner_extraction(text):
    # Set up client to interact with Ollama model
    client = Client(
        host='http://localhost:11434',
        headers={'x-some-header': 'some-value'}
    )
    response = client.chat(model='llama3.2', messages=[
        {"role": "user", "content": f"Extract named entities from: {text}"}
    ])
    return response['message']['content']
```

```python
import gradio as gr
```

```python
# Define Gradio UI
interface = gr.Interface(
    fn=ner_extraction,
    inputs=gr.Textbox(lines=5, placeholder="Enter text here ..."),
    outputs="text",
    title="Named Entity Recognition App",
    description="Enter text and see extracted named entities using Ollama."
)
```
